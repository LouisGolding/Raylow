````python
import os
import time
import textwrap
from typing import Generator
from dotenv import load_dotenv


import faiss
from sentence_transformers import CrossEncoder
from llama_index.core import (
    VectorStoreIndex,
    StorageContext,
    load_index_from_storage,
    PromptTemplate,
    Document
)
from llama_index.core.schema import TextNode
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
from llama_index.core.node_parser import SentenceSplitter
from llama_index.vector_stores.faiss import FaissVectorStore
from llama_index.readers.file import PyMuPDFReader
from llama_index.readers.web import SimpleWebPageReader
from llama_index.llms.mistralai import MistralAI
from llama_index.core import PromptTemplate
from llama_index.llms.openai import OpenAI

import json


from llama_index.core.node_parser import (
    SentenceSplitter,
    SemanticSplitterNodeParser,
)
from docling.document_converter import DocumentConverter


````

````python


load_dotenv()

class VectorDBBuilder:
    def __init__(self):
        self.config = {
            "embedding_model": "sentence-transformers/all-MiniLM-L6-v2",
            "chunk_size": 1000,
            "chunk_overlap": 100,
            "vector_db_dir": "./faiss_semantic_chunking",
            "sources": [
                "Regulations/CSRD.pdf",
        
            ]
        }
        self.embed_model = HuggingFaceEmbedding(
            model_name=self.config["embedding_model"]
        )

    def load_documents(self):
        """Load documents using Docling"""
        documents = []
        loader = DocumentConverter()
        for source in self.config["sources"]:

            try: 
                result = loader.convert(source)

                # Get content as markdown
                markdown_content = result.document.export_to_markdown()
                print(f"\n📄 Document Content from {source}:")
                print(f"{markdown_content[:500]}...\n")  # Print first 500 chars
                
                # Create a Document object with the markdown content. We need a llama-index Document for compatibility with our RAG pipeline. Not Docling document objects.
                doc = Document(
                    text=markdown_content, 
                    metadata={"source": source}
                )
                documents.append(doc)
                print(f"✅ Loaded document with Docling: {source}")
            
            except Exception as e: 
                print(f"❌ Error loading document {source}: {e}")

        return documents

    def save_nodes_to_file(self, nodes, output_path="chunks.txt"):
        """Save all nodes to a text file with clear separation"""
        print(f"Saving {len(nodes)} chunks to {output_path}...")
        
        # Group nodes by source
        source_to_nodes = {}
        for node in nodes:
            source = node.metadata.get("source", "Unknown")
            if source not in source_to_nodes:
                source_to_nodes[source] = []
            source_to_nodes[source].append(node)
        
        # Write to file
        with open(output_path, "w", encoding="utf-8") as outfile:
            outfile.write("=" * 100 + "\n")
            outfile.write(f"ALL SEMANTIC CHUNKS\n")
            outfile.write(f"Total chunks: {len(nodes)}\n")
            outfile.write("=" * 100 + "\n\n")
            
            chunk_counter = 1
            
            # Process each source
            for source, source_nodes in source_to_nodes.items():
                outfile.write("\n" + "=" * 100 + "\n")
                outfile.write(f"SOURCE: {source}\n")
                outfile.write("=" * 100 + "\n\n")
                
                # Write each node for this source
                for node in source_nodes:
                    text = node.text.strip()
                    node_id = node.node_id if hasattr(node, 'node_id') else f"node_{chunk_counter}"
                    
                    # Write node header
                    outfile.write("#" * 100 + "\n")
                    outfile.write(f"CHUNK {chunk_counter} | ID: {node_id}\n")
                    
                    # If node has start/end position info, include it
                    if hasattr(node, 'start_char_idx') and hasattr(node, 'end_char_idx'):
                        outfile.write(f"Character Range: {node.start_char_idx} - {node.end_char_idx} | ")
                    
                    outfile.write(f"Length: {len(text)} characters\n")
                    outfile.write("-" * 100 + "\n\n")
                    
                    # Write the text
                    outfile.write(text)
                    
                    # Add separator after node content
                    outfile.write("\n\n" + "-" * 100 + "\n\n")
                    
                    chunk_counter += 1
        
        print(f"✅ Successfully wrote all chunks to: {output_path}")
        return output_path

    def build_vector_store(self, save_chunks_to_file=True):
        """Build and save FAISS vector store using proper LlamaIndex persistence"""
        # Load and process documents
        documents = self.load_documents()
        print(f"📚 Loaded {len(documents)} document(s)")

        # Implement semantic chunking
        text_parser = SemanticSplitterNodeParser(
            buffer_size=1, 
            breakpoint_percentile_threshold=75, 
            embed_model=self.embed_model
        )
            
        # Get nodes from documents - this handles all the text splitting
        nodes = text_parser.get_nodes_from_documents(documents)
        print(f"✂️  Split into {len(nodes)} chunks")
        
        # Save nodes to file if requested
        if save_chunks_to_file:
            chunks_path = "all_chunks.txt"
            self.save_nodes_to_file(nodes, output_path=chunks_path)
        
        # Display a sample chunk for debugging
        if nodes:
            print(f"📌 Chunk {min(30, len(nodes)-1)}: {nodes[min(30, len(nodes)-1)].text[:500]}...")

        # Create FAISS index and storage context
        d = 384  # Dimension of all-MiniLM-L6-v2 embeddings
        faiss_index = faiss.IndexFlatL2(d)
        vector_store = FaissVectorStore(faiss_index=faiss_index)
        storage_context = StorageContext.from_defaults(vector_store=vector_store)
        
        # Create index with nodes and embeddings
        index = VectorStoreIndex(
            nodes,
            storage_context=storage_context,
            embed_model=self.embed_model
        )

        # Persist to disk
        os.makedirs(self.config["vector_db_dir"], exist_ok=True)
        storage_context.persist(persist_dir=self.config["vector_db_dir"])
        print(f"🚀 FAISS Vector DB built and saved to {self.config['vector_db_dir']}")

# Standalone function to just get and save chunks without building the vector store
def extract_and_save_chunks(output_path="all_chunks.txt"):
    """Extract chunks from documents and save them to a file without building the vector store"""
    builder = VectorDBBuilder()
    documents = builder.load_documents()
    print(f"📚 Loaded {len(documents)} document(s)")
    
    text_parser = SemanticSplitterNodeParser(
        buffer_size=1, 
        breakpoint_percentile_threshold=75, 
        embed_model=builder.embed_model
    )
    
    nodes = text_parser.get_nodes_from_documents(documents)
    print(f"✂️  Split into {len(nodes)} chunks")
    
    builder.save_nodes_to_file(nodes, output_path=output_path)
    return output_path


if __name__ == "__main__":
    # Build vector store and save chunks
    builder = VectorDBBuilder()
    builder.build_vector_store(save_chunks_to_file=True)
    
 
````

````python
import json
import os

def print_chunk(chunk_number=50):
    """Print a specific chunk from the saved docstore"""
    # Path to the docstore file
    docstore_path = "./faiss_semantic_chunking/docstore.json"
    
    # Load the docstore
    with open(docstore_path, "r") as f:
        docstore_data = json.load(f)
    
    # Get all nodes and sort them
    nodes = []
    for node_id, node_data in docstore_data.get("docstore/data", {}).items():
        if "__data__" in node_data:
            data = node_data["__data__"]
            text = data.get("text", "")
            source = data.get("metadata", {}).get("source", "Unknown")
            start_idx = data.get("start_char_idx", 0)
            
            nodes.append({
                "id": node_id,
                "text": text,
                "source": source,
                "position": start_idx
            })
    
    # Sort nodes by source and position
    nodes.sort(key=lambda x: (x["source"], x["position"]))
    
    # Check if the requested chunk exists
    if chunk_number <= 0 or chunk_number > len(nodes):
        print(f"Error: Chunk {chunk_number} doesn't exist. Valid range is 1-{len(nodes)}")
        return
    
    # Get the requested chunk (adjust for 0-based indexing)
    chunk = nodes[chunk_number-1]
    
    # Print the chunk
    print("\n" + "=" * 80)
    print(f"CHUNK {chunk_number}")
    print(f"ID: {chunk['id']}")
    print(f"Source: {chunk['source']}")
    print(f"Length: {len(chunk['text'])} characters")
    print("=" * 80)
    print("\n" + chunk["text"] + "\n")
    print("=" * 80)

# Run the function to print chunk 15
print_chunk(60)
````

````python
load_dotenv()


class SustainabilityExpert:
    

    def __init__(self):
        self.vector_db_dir = "./faiss_semantic_chunking"
        self.embed_model = HuggingFaceEmbedding(
            model_name="sentence-transformers/all-MiniLM-L6-v2"
        )
        
        # Load FAISS vector store from persistence
        self.vector_store = FaissVectorStore.from_persist_dir(self.vector_db_dir)
        self.storage_context = StorageContext.from_defaults(
            vector_store=self.vector_store,
            persist_dir=self.vector_db_dir
        )
        
        # Load index
        self.index = load_index_from_storage(
            storage_context=self.storage_context,
            embed_model=self.embed_model  
)
        
        # Configure retriever
        self.retriever = self.index.as_retriever(similarity_top_k=10)    # .as_retriever method sets up the index so it can fetch relevant text chunks based on how similar they are to a given query. It's like preparing a librarian (the retriever) to find books (text chunks) that match your request (query).


        self.re_ranker = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-6-v2")
        
        # Configure Mistral
        self.llm = MistralAI(
            model="mistral-large-latest",
            temperature=0,
            api_key=os.environ["MISTRAL_API_KEY"]
        )

        

        
        # Define prompt template
        self.qa_prompt = PromptTemplate("""\
You're a sustainability regulations expert. Answer clearly using the context.

Context: {context}
Question: {question}

Help users comply with regulations for their companies. Your aim is to help them and be as clear as possible. If unsure, say 'I don't know'. Answer:\
""")

    def format_response(self, text: str, width: int = 80) -> str:
        """Format response with proper line wrapping"""
        return textwrap.fill(text, width=width)

    def rerank_documents(self, query: str, nodes: list) -> list:
        """Re-rank nodes using cross-encoder"""
        node_texts = [node.text for node in nodes]
        scores = self.re_ranker.predict([(query, text) for text in node_texts])
        scored_nodes = sorted(zip(nodes, scores), key=lambda x: x[1], reverse=True)
        return [node for node, _ in scored_nodes[:5]]  # Return top 5

    def query(self, question: str) -> Generator[str, None, None]:
        """Streaming query response"""
        # Retrieve and rerank nodes
        nodes = self.retriever.retrieve(question)  # '.retrieve()' uses the embedding model to convert the query into a vector, searches the vector store for similar vectors, and returns the associated text chunks.
        reranked_nodes = self.rerank_documents(question, nodes)
        
        # Build context
        context = "\n\n".join([n.text for n in reranked_nodes])
        
        # Stream response
        response = self.llm.stream_complete(
            self.qa_prompt.format(context=context, question=question)
        )
        
        for chunk in response:
            yield chunk.delta or ""

if __name__ == "__main__":
    expert = SustainabilityExpert()
    print("🌱 Sustainability Regulations Query System")
    print("Type 'q', 'exit', or 'quit' to end the session.\n")

    while True:
        question = input("Ask your question:\n> ").strip()
        if question.lower() in ('q', 'exit', 'quit'):
            print("\n👋 Goodbye!")
            break

        start_time = time.time()
        print("\n🔍 Searching regulations...")

        # Stream the response
        print("\n📝 Answer:")
        response = ""
        for chunk in expert.query(question):
            print(chunk, end="", flush=True)
            response += chunk

        print(f"\n\n⏱️  Response time: {time.time() - start_time:.2f}s\n")
````

````python
import json
import textwrap
import os
import time
from typing import Dict, List, Any, Union

from dotenv import load_dotenv
from llama_index.core.retrievers import VectorIndexRetriever
from llama_index.core.query_engine import RetrieverQueryEngine
from llama_index.core.indices import VectorStoreIndex
from llama_index.core.storage import StorageContext
from llama_index.core import load_index_from_storage
from llama_index.vector_stores.faiss import FaissVectorStore
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
from llama_index.llms.openai import OpenAI
from llama_index.core.prompts import PromptTemplate
from sentence_transformers import CrossEncoder

load_dotenv()


class SustainabilityExpert:
    
    def __init__(self):
        self.vector_db_dir = "./faiss_semantic_chunking"
        self.embed_model = HuggingFaceEmbedding(
            model_name="sentence-transformers/all-MiniLM-L6-v2"
        )
        
        # Load FAISS vector store from persistence
        self.vector_store = FaissVectorStore.from_persist_dir(self.vector_db_dir)
        self.storage_context = StorageContext.from_defaults(
            vector_store=self.vector_store,
            persist_dir=self.vector_db_dir
        )
        
        # Load index
        self.index = load_index_from_storage(
            storage_context=self.storage_context,
            embed_model=self.embed_model  
        )
        
        # Configure retriever - reduced from top 10 to top 5
        self.retriever = self.index.as_retriever(similarity_top_k=5)

        self.re_ranker = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-6-v2")
        
        # Configure OpenAI with JSON format
        self.llm = OpenAI(
            model="gpt-4o",
            temperature=0,
            api_key=os.environ["OPENAI_API_KEY"], 
            response_format={"type": "json_object"} 
        )
        
        # Define prompt template with explicit format instructions
        self.qa_prompt = PromptTemplate("""\
        You're a sustainability regulations expert. Answer clearly using the context.

        Context: {context}
        Question: {question}
                                                
        Help users comply with regulations for their companies. Your aim is to help them and be as clear as possible. If unsure, say 'I don't know'.

        YOUR RESPONSE FORMAT IS CRITICAL:
        Return ONLY a simple JSON object with EXACTLY ONE KEY called "text".
        The value of "text" should be your answer as a plain string.

        Example of CORRECT format:
        {{"text": "The CSRD requires companies to report on their sustainability practices."}}

        DO NOT use nested JSON objects.
        DO NOT use markdown formatting.
        DO NOT wrap your response in code blocks.
        DO NOT include fields like "answer" or any other field names.
        """)
        
        # Flag to control whether to return full RAGAS format
        self.return_full_format = False

    def format_response(self, text: str, width: int = 80) -> str:
        """Format response with proper line wrapping"""
        return textwrap.fill(text, width=width)

    def rerank_documents(self, query: str, nodes: list) -> list:
        """Re-rank nodes using cross-encoder"""
        node_texts = [node.text for node in nodes]
        scores = self.re_ranker.predict([(query, text) for text in node_texts])
        scored_nodes = sorted(zip(nodes, scores), key=lambda x: x[1], reverse=True)
        # Return top 2 nodes (reduced from 5)
        return [node for node, _ in scored_nodes[:2]]
    
    def query(self, question: str) -> Dict:
        """
        Query with RAGAS-compatible JSON response format
        
        Args:
            question: The user's question about sustainability regulations
                
        Returns:
            If self.return_full_format is True:
                Returns complete RAGAS format with all required fields
            Otherwise:
                Returns simple {"text": answer_text} format
        """
        # Retrieve and rerank nodes
        nodes = self.retriever.retrieve(question)
        # Fixed: Use 'question' variable instead of undefined 'query'
        reranked_nodes = self.rerank_documents(question, nodes)
        
        # Get context texts for RAGAS format
        context_texts = [node.text for node in reranked_nodes]
        
        # Build context for LLM
        context = "\n\n".join(context_texts)
        
        # Get response as JSON
        response = self.llm.complete(
            self.qa_prompt.format(context=context, question=question)
        )
        
        # Get the raw text from the response
        if hasattr(response, 'text'):
            raw_text = response.text
        else:
            raw_text = str(response)
        
        # Parse the JSON to extract the answer text
        try:
            # Parse the JSON
            parsed_json = json.loads(raw_text)
            
            # Check if it contains a 'text' field - use that as the answer
            if "text" in parsed_json:
                answer_text = parsed_json["text"]
            # If it contains an 'answer' field, use that instead
            elif "answer" in parsed_json:
                answer_text = parsed_json["answer"]
            # If the response is just a string or doesn't have the expected fields
            else:
                # Use the first field we find or the whole object stringified
                first_value = next(iter(parsed_json.values()), raw_text) 
                if isinstance(first_value, str):
                    answer_text = first_value
                else:
                    answer_text = json.dumps(first_value)
                
        except json.JSONDecodeError:
            # If the response is not valid JSON (shouldn't happen with response_format=json_object)
            # Clean up any markdown code blocks if present
            if raw_text.startswith("```"):
                raw_text = raw_text.replace("```json", "", 1).lstrip()
                if "```" in raw_text:
                    raw_text = raw_text.rsplit("```", 1)[0].rstrip()
            
            # Use cleaned raw text as fallback
            answer_text = raw_text
        
        # Return format based on flag
        if self.return_full_format:
            # Build the EXACT RAGAS-compatible structure
            return {
                "question": question,
                "contexts": context_texts,
                "answer": answer_text,
                # Note: ground_truth will be added by the evaluator
            }
        else:
            # Return simple format for normal usage
            return {"text": answer_text}


if __name__ == "__main__":
    expert = SustainabilityExpert()
    print("🌱 Sustainability Regulations Query System")
    print("Type 'q', 'exit', or 'quit' to end the session.\n")

    while True:
        question = input("Ask your question:\n> ").strip()
        if question.lower() in ('q', 'exit', 'quit'):
            print("\n👋 Goodbye!")
            break

        start_time = time.time()
        print("\n🔍 Searching regulations...")

        # Get response
        response = expert.query(question)
        
        # Pretty print the response
        print("\n📝 Answer:")
        if "text" in response:
            print(expert.format_response(response["text"]))
        else:
            print(json.dumps(response, indent=2))
        
        print(f"\n⏱️  Response time: {time.time() - start_time:.2f}s\n")
````
